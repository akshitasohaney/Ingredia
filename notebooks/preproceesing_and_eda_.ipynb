```python
import re
import pandas as pd

mapping_review = name_map.copy()

mapping_review["raw_name_clean"] = (
    mapping_review["raw_name"]
    .fillna("")
    .astype(str)
    .str.strip()
)

patterns = {
    "manufacturer_code": r"\bF\.?I\.?L\.?\b|F\d{2}|B\d{4,}",
    "may_contain_statement": r"may contain|peut contenir|may contain/peut",
    "label_metadata": (
        r"ingredients?\s*:|inactive ingredients|"
        r"activator|emulsion|bonding treatment"
    ),
    "possible_multiple_ingredients": r"\s[|]\s|;",
    "possible_product_text": (
        r"www\.|fabricado|exportado|duschgel|"
        r"cr[eè]me colorante|naturkosmetik"
    )
}

for flag, pattern in patterns.items():
    mapping_review[flag] = mapping_review["raw_name_clean"].str.contains(
        pattern, case=False, regex=True, na=False
    )

flag_columns = list(patterns.keys())

mapping_review["needs_manual_review"] = mapping_review[
    flag_columns
].any(axis=1)

print("Mappings requiring review:",
      int(mapping_review["needs_manual_review"].sum()))

display(
    mapping_review.loc[
        mapping_review["needs_manual_review"],
        [
            "raw_name", "canonical_name", "method",
            "confidence", "ingredient_id"
        ] + flag_columns
    ].head(60)
)

mapping_review.to_csv(
    OUTPUT_DIR / "ingredient_mapping_review.csv",
    index=False
)
```

```python
import pandas as pd
from pathlib import Path

# Reuse mapping_review from Cell 10.
# This creates a review file; it does not alter your source data.

review_queue = mapping_review[
    mapping_review["needs_manual_review"]
].copy()

def recommend_review(row):
    method = str(row.get("method", "")).lower()
    confidence = pd.to_numeric(
        pd.Series([row.get("confidence")]), errors="coerce"
    ).iloc[0]

    if row.get("manufacturer_code", False):
        return "Separate manufacturer code from ingredient identity"

    if row.get("may_contain_statement", False):
        return "Review optional colorants and may-contain context"

    if row.get("possible_product_text", False):
        return "Separate product description from ingredient text"

    if row.get("possible_multiple_ingredients", False):
        return "Verify ingredient splitting and each resulting mapping"

    if method in {"unresolved", "unresolved_residual"}:
        return "Manual identification required; do not auto-map"

    if pd.isna(confidence) or confidence < 0.8:
        return "Verify canonical ingredient against a trusted source"

    return "Review context before accepting mapping"

review_queue["recommended_action"] = review_queue.apply(
    recommend_review, axis=1
)

review_queue["review_status"] = "pending"

# Put low-confidence and unresolved mappings first.
review_queue["confidence_num"] = pd.to_numeric(
    review_queue["confidence"], errors="coerce"
)

review_queue = review_queue.sort_values(
    ["confidence_num", "method"],
    ascending=[True, True],
    na_position="first"
)

review_columns = [
    "raw_name",
    "canonical_name",
    "method",
    "confidence",
    "ingredient_id",
    "part_index",
    "recommended_action",
    "review_status",
]

review_columns = [
    col for col in review_columns if col in review_queue.columns
]

review_path = OUTPUT_DIR / "ingredient_mapping_review_queue.csv"

review_queue[review_columns].to_csv(
    review_path, index=False
)

print("Review queue rows:", len(review_queue))
print("Saved to:", review_path)
display(review_queue[review_columns].head(20))
```

```python

import sqlite3
import pandas as pd
from pathlib import Path

# ---------- Paths ----------
DATA_DIR = Path(
    "/content/drive/MyDrive/ingredia_backup/ingredia_data/cleaned"
)
DB_PATH = DATA_DIR / "ingredia.db"

csv_tables = {
    "ingredients": "api_ingredients.csv",
    "products": "api_products.csv",
    "product_ingredients": "api_product_ingredients.csv",
    "ingredient_aliases": "api_ingredient_aliases.csv",
}

# ---------- Create database ----------
if DB_PATH.exists():
    DB_PATH.unlink()  # Rebuild only this generated database file

conn = sqlite3.connect(DB_PATH)

try:
    expected_counts = {}

    for table_name, filename in csv_tables.items():
        csv_path = DATA_DIR / filename
        df = pd.read_csv(csv_path)

        df.to_sql(
            table_name,
            conn,
            if_exists="replace",
            index=False
        )

        expected_counts[table_name] = len(df)

    # Add indexes to speed up common searches and joins.
    conn.executescript("""
        CREATE INDEX idx_ingredients_name
        ON ingredients(inci_name);

        CREATE INDEX idx_products_name
        ON products(name);

        CREATE INDEX idx_products_barcode
        ON products(barcode_ean);

        CREATE INDEX idx_product_ingredients_product
        ON product_ingredients(product_id);

        CREATE INDEX idx_product_ingredients_ingredient
        ON product_ingredients(ingredient_id);

        CREATE INDEX idx_aliases_ingredient
        ON ingredient_aliases(ingredient_id);
    """)

    conn.commit()

    # ---------- Verify database ----------
    print("Database:", DB_PATH)
    print("\nRow-count verification:")

    for table_name, expected in expected_counts.items():
        actual = conn.execute(
            f"SELECT COUNT(*) FROM {table_name}"
        ).fetchone()[0]

        print(f"{table_name}: {actual} rows (CSV: {expected})")
        assert actual == expected, f"Row count mismatch: {table_name}"

    # Test a sample product-to-ingredient join.
    sample = pd.read_sql_query("""
        SELECT
            p.product_id,
            p.name AS product_name,
            i.ingredient_id,
            i.inci_name,
            pi.position_index
        FROM products p
        JOIN product_ingredients pi
            ON p.product_id = pi.product_id
        JOIN ingredients i
            ON pi.ingredient_id = i.ingredient_id
        ORDER BY p.product_id, pi.position_index
        LIMIT 10
    """, conn)

    print("\nSample product-ingredient join:")
    display(sample)

    print("\nDatabase verification passed.")

finally:
    conn.close()
```

```python

from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/ingredia_backup/ingredia_data"
)
CLEANED_DIR = PROJECT_DIR / "cleaned"
DB_PATH = CLEANED_DIR / "ingredia.db"

MAIN_PY = PROJECT_DIR / "main.py"

api_code = r'''
import os
import sqlite3
from pathlib import Path
from typing import Optional

from fastapi import FastAPI, HTTPException, Query

DB_PATH = Path(
    os.environ.get(
        "INGREDIA_DB_PATH",
        "/content/drive/MyDrive/ingredia_backup/ingredia_data/cleaned/ingredia.db"
    )
)

app = FastAPI(
    title="Ingredia API",
    description=(
        "Ingredient and cosmetic product information API. "
        "Source-reported information is not a universal safety verdict."
    ),
    version="0.1.0"
)


def get_connection():
    if not DB_PATH.exists():
        raise RuntimeError(f"Database not found: {DB_PATH}")

    conn = sqlite3.connect(str(DB_PATH))
    conn.row_factory = sqlite3.Row
    conn.execute("PRAGMA query_only = ON")
    return conn


def rows_to_dicts(rows):
    return [dict(row) for row in rows]


@app.get("/")
def root():
    return {
        "name": "Ingredia API",
        "status": "running",
        "docs": "/docs"
    }


@app.get("/health")
def health():
    try:
        with get_connection() as conn:
            count = conn.execute(
                "SELECT COUNT(*) FROM ingredients"
            ).fetchone()[0]
        return {"status": "healthy", "ingredient_records": count}
    except Exception:
        raise HTTPException(
            status_code=503,
            detail="Database is unavailable or not initialized"
        )


@app.get("/ingredients")
def search_ingredients(
    search: Optional[str] = Query(default=None, max_length=100),
    limit: int = Query(default=20, ge=1, le=100),
    offset: int = Query(default=0, ge=0)
):
    with get_connection() as conn:
        if search and search.strip():
            term = "%" + search.strip() + "%"
            rows = conn.execute(
                """
                SELECT ingredient_id, inci_name, functions,
                       chemical_description, cosing_matched,
                       cosing_ref_no, cas_number, ec_number
                FROM ingredients
                WHERE inci_name LIKE ?
                ORDER BY inci_name
                LIMIT ? OFFSET ?
                """,
                (term, limit, offset)
            ).fetchall()
        else:
            rows = conn.execute(
                """
                SELECT ingredient_id, inci_name, functions,
                       chemical_description, cosing_matched,
                       cosing_ref_no, cas_number, ec_number
                FROM ingredients
                ORDER BY inci_name
                LIMIT ? OFFSET ?
                """,
                (limit, offset)
            ).fetchall()

    return {
        "count": len(rows),
        "limit": limit,
        "offset": offset,
        "results": rows_to_dicts(rows)
    }


@app.get("/ingredients/{ingredient_id}")
def get_ingredient(ingredient_id: int):
    with get_connection() as conn:
        row = conn.execute(
            "SELECT * FROM ingredients WHERE ingredient_id = ?",
            (ingredient_id,)
        ).fetchone()

    if row is None:
        raise HTTPException(status_code=404, detail="Ingredient not found")

    return dict(row)


@app.get("/products")
def search_products(
    search: Optional[str] = Query(default=None, max_length=150),
    barcode: Optional[str] = Query(default=None, max_length=50),
    limit: int = Query(default=20, ge=1, le=100),
    offset: int = Query(default=0, ge=0)
):
    clauses = []
    params = []

    if search and search.strip():
        clauses.append("name LIKE ?")
        params.append("%" + search.strip() + "%")

    if barcode and barcode.strip():
        clauses.append("barcode_ean = ?")
        params.append(barcode.strip())

    sql = "SELECT * FROM products"
    if clauses:
        sql += " WHERE " + " AND ".join(clauses)

    sql += " ORDER BY name LIMIT ? OFFSET ?"
    params.extend([limit, offset])

    with get_connection() as conn:
        rows = conn.execute(sql, params).fetchall()

    return {
        "count": len(rows),
        "limit": limit,
        "offset": offset,
        "results": rows_to_dicts(rows)
    }


@app.get("/products/{product_id}/ingredients")
def get_product_ingredients(product_id: int):
    with get_connection() as conn:
        product = conn.execute(
            "SELECT * FROM products WHERE product_id = ?",
            (product_id,)
        ).fetchone()

        if product is None:
            raise HTTPException(
                status_code=404, detail="Product not found"
            )

        rows = conn.execute(
            """
            SELECT i.*, pi.position_index,
                   pi.concentration_percentage
            FROM product_ingredients pi
            JOIN ingredients i
              ON i.ingredient_id = pi.ingredient_id
            WHERE pi.product_id = ?
            ORDER BY pi.position_index
            """,
            (product_id,)
        ).fetchall()

    return {
        "product": dict(product),
        "ingredient_count": len(rows),
        "ingredients": rows_to_dicts(rows)
    }
'''

MAIN_PY.write_text(api_code, encoding="utf-8")

print("Created:", MAIN_PY)
print("Database exists:", DB_PATH.exists())
```

```python

from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/ingredia_backup/ingredia_data"
)
MAIN_PY = PROJECT_DIR / "main.py"

alias_search_code = r'''

@app.get("/ingredient-search")
def ingredient_search(
    search: str = Query(..., min_length=1, max_length=100),
    limit: int = Query(default=20, ge=1, le=100),
    offset: int = Query(default=0, ge=0)
):
    term = search.strip()
    if not term:
        raise HTTPException(status_code=422, detail="Search cannot be blank")

    pattern = "%" + term + "%"
    prefix = term + "%"

    with get_connection() as conn:
        # Discover alias columns from the actual database schema.
        alias_columns = [
            row["name"]
            for row in conn.execute(
                "PRAGMA table_info(ingredient_aliases)"
            ).fetchall()
        ]

        if "ingredient_id" not in alias_columns:
            raise HTTPException(
                status_code=500,
                detail="Alias table has no ingredient_id column"
            )

        possible_name_columns = [
            "raw_name", "canonical_name", "inci_name",
            "alias", "name", "normalized_name"
        ]
        name_columns = [
            col for col in possible_name_columns
            if col in alias_columns
        ]

        alias_ids = set()

        if name_columns:
            conditions = " OR ".join(
                f'CAST("{col}" AS TEXT) LIKE ?'
                for col in name_columns
            )
            params = [pattern] * len(name_columns)

            rows = conn.execute(
                f"""
                SELECT DISTINCT ingredient_id
                FROM ingredient_aliases
                WHERE {conditions}
                """,
                params
            ).fetchall()

            alias_ids = {
                row["ingredient_id"] for row in rows
                if row["ingredient_id"] is not None
            }

        # Search direct INCI names, plus ingredients reached through aliases.
        direct_rows = conn.execute(
            """
            SELECT ingredient_id, inci_name, functions,
                   chemical_description, cosing_matched,
                   cosing_ref_no, cas_number, ec_number
            FROM ingredients
            WHERE inci_name LIKE ?
            """,
            (pattern,)
        ).fetchall()

        candidates = {
            row["ingredient_id"]: dict(row)
            for row in direct_rows
        }

        if alias_ids:
            valid_ids = [
                ingredient_id for ingredient_id in alias_ids
            ]
            placeholders = ",".join("?" for _ in valid_ids)

            alias_rows = conn.execute(
                f"""
                SELECT ingredient_id, inci_name, functions,
                       chemical_description, cosing_matched,
                       cosing_ref_no, cas_number, ec_number
                FROM ingredients
                WHERE ingredient_id IN ({placeholders})
                """,
                valid_ids
            ).fetchall()

            for row in alias_rows:
                candidates[row["ingredient_id"]] = dict(row)

        # Rank exact name matches, then prefix matches, then other matches.
        def rank(item):
            name = str(item["inci_name"] or "").casefold()
            query = term.casefold()

            if name == query:
                match_rank = 0
            elif name.startswith(query):
                match_rank = 1
            elif item["ingredient_id"] in alias_ids:
                match_rank = 2
            else:
                match_rank = 3

            return (match_rank, name)

        results = sorted(candidates.values(), key=rank)
        total = len(results)
        page = results[offset:offset + limit]

    return {
        "query": term,
        "count": len(page),
        "total_matches": total,
        "limit": limit,
        "offset": offset,
        "results": page,
        "note": (
            "Alias matches are search suggestions, not confirmation "
            "that the raw label text is a verified synonym."
        )
    }
'''

# Avoid accidentally adding the same endpoint twice.
current_code = MAIN_PY.read_text(encoding="utf-8")

if '@app.get("/ingredient-search")' in current_code:
    print("The alias-aware endpoint already exists; no changes made.")
else:
    with MAIN_PY.open("a", encoding="utf-8") as f:
        f.write("\n" + alias_search_code + "\n")
    print("Added /ingredient-search to:", MAIN_PY)
```
